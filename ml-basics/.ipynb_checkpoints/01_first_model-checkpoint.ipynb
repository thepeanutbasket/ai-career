```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
data = load_iris(as_frame=True)
df = data.frame
df.head()
```

```python
X = df.drop(columns="target")
y = df["target"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=40
)
```

```python
model = LogisticRegression(max_iter=200)
model.fit(X_train, y_train)
pred = model.predict(X_test)
accuracy_score(y_test, pred)
```

load_iris(): 
Loads the iris data.

as_frame=True: 
Loads the data in a dataframe format.

X: 
Features used to make predictions.

y: 
The target the model is trying to predict.

train_test_split: 
Separates data used for training from unseen data used for evaluation.

model.fit(): 
Learns patterns from the training data.

model.predict(): 
Uses the trained model to predict new data.

**Review**

What is a feature?
- A feature is an input variable used by the model to make predictions.

What is a target?
- A target is the value that the model is trying to predict.

What does model.fit() do?
- model.fit() trains the model by learning patterns and relationships from the training data.

Why do we split data into training and test sets?
- We need a training set to train the model, and we need test sets so we have fresh data to the model to evaluate the performance.

What does model.predict() do?
- model.predict() uses a trained model to make predictions on new input data.
